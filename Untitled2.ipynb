```python
# ============================================================
# STEP 3E — ROUTE OVERLAP / COVERAGE ANALYSIS
# ============================================================

print("=" * 70)
print("V2 ROUTE OVERLAP ANALYSIS")
print("=" * 70)

# ------------------------------------------------------------
# S1 coverage by route
# ------------------------------------------------------------

coverage = con.execute(f"""
WITH
a AS (
    SELECT DISTINCT source1_entity_id
    FROM read_parquet('{route_a}')
),
b AS (
    SELECT DISTINCT source1_entity_id
    FROM read_parquet('{route_b}')
),
c AS (
    SELECT DISTINCT source1_entity_id
    FROM read_parquet('{route_c}')
)
SELECT
    (SELECT COUNT(*) FROM a) AS route_a_s1,
    (SELECT COUNT(*) FROM b) AS route_b_s1,
    (SELECT COUNT(*) FROM c) AS route_c_s1,

    (
        SELECT COUNT(*)
        FROM b
        WHERE source1_entity_id NOT IN
              (SELECT source1_entity_id FROM a)
    ) AS b_new_vs_a,

    (
        SELECT COUNT(*)
        FROM c
        WHERE source1_entity_id NOT IN
              (SELECT source1_entity_id FROM a)
    ) AS c_new_vs_a,

    (
        SELECT COUNT(*)
        FROM (
            SELECT source1_entity_id FROM a
            UNION
            SELECT source1_entity_id FROM b
            UNION
            SELECT source1_entity_id FROM c
        )
    ) AS union_s1
""").fetchdf()

print("\nS1 COVERAGE:")
print(coverage.to_string(index=False))


# ------------------------------------------------------------
# Pair overlap
# ------------------------------------------------------------

pair_overlap = con.execute(f"""
WITH
a AS (
    SELECT DISTINCT source1_entity_id, matched_entity_id
    FROM read_parquet('{route_a}')
),
b AS (
    SELECT DISTINCT source1_entity_id, matched_entity_id
    FROM read_parquet('{route_b}')
),
c AS (
    SELECT DISTINCT source1_entity_id, matched_entity_id
    FROM read_parquet('{route_c}')
)
SELECT
    (SELECT COUNT(*) FROM a) AS A_pairs,
    (SELECT COUNT(*) FROM b) AS B_pairs,
    (SELECT COUNT(*) FROM c) AS C_pairs,

    (
        SELECT COUNT(*)
        FROM b
        WHERE (source1_entity_id, matched_entity_id)
              NOT IN (
                  SELECT source1_entity_id, matched_entity_id
                  FROM a
              )
    ) AS B_new_pairs_vs_A,

    (
        SELECT COUNT(*)
        FROM c
        WHERE (source1_entity_id, matched_entity_id)
              NOT IN (
                  SELECT source1_entity_id, matched_entity_id
                  FROM a
              )
    ) AS C_new_pairs_vs_A,

    (
        SELECT COUNT(*)
        FROM c
        WHERE (source1_entity_id, matched_entity_id)
              NOT IN (
                  SELECT source1_entity_id, matched_entity_id
                  FROM a
                  UNION
                  SELECT source1_entity_id, matched_entity_id
                  FROM b
              )
    ) AS C_new_pairs_vs_A_B
""").fetchdf()

print("\nPAIR OVERLAP:")
print(pair_overlap.to_string(index=False))


# ------------------------------------------------------------
# Candidate union size
# ------------------------------------------------------------

union_count = con.execute(f"""
SELECT COUNT(*)
FROM (
    SELECT source1_entity_id, matched_entity_id
    FROM read_parquet('{route_a}')

    UNION

    SELECT source1_entity_id, matched_entity_id
    FROM read_parquet('{route_b}')

    UNION

    SELECT source1_entity_id, matched_entity_id
    FROM read_parquet('{route_c}')
)
""").fetchone()[0]

print(f"\nA+B+C unique candidate pairs: {union_count:,}")

print("\n" + "=" * 70)
print("OVERLAP ANALYSIS COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# STEP 3G — ACTUAL SCRIPT COMPOSITION OF UNCOVERED NAMES
# Corrected Python version — no SQL Unicode regex
# ============================================================

import unicodedata

print("=" * 70)
print("UNCOVERED S1 — ACTUAL SCRIPT ANALYSIS")
print("=" * 70)

# Get uncovered names from DuckDB
uncovered_names = con.execute(f"""
WITH covered AS (
    SELECT DISTINCT source1_entity_id
    FROM (
        SELECT source1_entity_id
        FROM read_parquet('{route_a}')

        UNION

        SELECT source1_entity_id
        FROM read_parquet('{route_b}')
    )
)
SELECT
    s1.entity_id,
    s1.business_name_clean
FROM source1 s1
LEFT JOIN covered c
    ON s1.entity_id = c.source1_entity_id
WHERE c.source1_entity_id IS NULL
""").fetchdf()


# ------------------------------------------------------------
# Unicode script classifier
# ------------------------------------------------------------

def get_script(text):

    if not text:
        return "Empty"

    scripts = set()

    for ch in str(text):

        code = ord(ch)

        # Devanagari
        if 0x0900 <= code <= 0x097F:
            scripts.add("Devanagari")

        # Bengali
        elif 0x0980 <= code <= 0x09FF:
            scripts.add("Bengali")

        # Gurmukhi
        elif 0x0A00 <= code <= 0x0A7F:
            scripts.add("Gurmukhi")

        # Gujarati
        elif 0x0A80 <= code <= 0x0AFF:
            scripts.add("Gujarati")

        # Tamil
        elif 0x0B80 <= code <= 0x0BFF:
            scripts.add("Tamil")

        # Telugu
        elif 0x0C00 <= code <= 0x0C7F:
            scripts.add("Telugu")

        # Kannada
        elif 0x0C80 <= code <= 0x0CFF:
            scripts.add("Kannada")

        # Malayalam
        elif 0x0D00 <= code <= 0x0D7F:
            scripts.add("Malayalam")

        # Arabic
        elif 0x0600 <= code <= 0x06FF:
            scripts.add("Arabic")

        # Hebrew
        elif 0x0590 <= code <= 0x05FF:
            scripts.add("Hebrew")

        # CJK
        elif 0x4E00 <= code <= 0x9FFF:
            scripts.add("CJK")

        # Cyrillic
        elif 0x0400 <= code <= 0x04FF:
            scripts.add("Cyrillic")

        # Greek
        elif 0x0370 <= code <= 0x03FF:
            scripts.add("Greek")

        # Ignore ASCII letters, digits, spaces and punctuation
        elif code < 128:
            continue

        else:
            # Other Unicode character
            category = unicodedata.category(ch)

            if category.startswith("L"):
                scripts.add("Other_NonLatin")

    if not scripts:
        return "Latin/ASCII"

    if len(scripts) == 1:
        return list(scripts)[0]

    return "Mixed/" + "+".join(sorted(scripts))


# ------------------------------------------------------------
# Classify names
# ------------------------------------------------------------

uncovered_names["script"] = uncovered_names[
    "business_name_clean"
].apply(get_script)

script_result = (
    uncovered_names["script"]
    .value_counts()
    .rename_axis("script")
    .reset_index(name="s1_count")
)

print("\nSCRIPT DISTRIBUTION:")
print(script_result.to_string(index=False))


# ------------------------------------------------------------
# Show examples from each script
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("EXAMPLES")
print("=" * 70)

for script in script_result["script"].head(15):

    examples = uncovered_names[
        uncovered_names["script"] == script
    ]["business_name_clean"].head(3).tolist()

    print(f"\n{script}:")
    for x in examples:
        print("  ", x)


print("\n" + "=" * 70)
print("SCRIPT ANALYSIS COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# STEP 3I — SMALL FUZZY BLOCKING EXPERIMENT
# ONLY 300 UNCOVERED S1 RECORDS
# ============================================================

from rapidfuzz import fuzz
from collections import defaultdict
import re
import pandas as pd
import numpy as np

print("=" * 70)
print("FUZZY BLOCKING EXPERIMENT — 300 UNCOVERED S1")
print("=" * 70)


# ------------------------------------------------------------
# 1. Prepare source names
# ------------------------------------------------------------

# names_df and uncovered_sample already exist from Step 3H

source_names = names_df["business_name_clean"].tolist()

print(f"Source names available: {len(source_names):,}")


# ------------------------------------------------------------
# 2. Blocking functions
# ------------------------------------------------------------

def first4_len(name):
    name = str(name)
    return (
        name[:4],
        len(name) // 5
    )


def first3_last3(name):
    name = str(name)
    return (
        name[:3],
        name[-3:] if len(name) >= 3 else name
    )


def trigrams(name):
    name = str(name)

    if len(name) < 3:
        return {name}

    return {
        name[i:i+3]
        for i in range(len(name) - 2)
    }


# ------------------------------------------------------------
# 3. Build block indexes
# ------------------------------------------------------------

print("\nBuilding blocking indexes...")

block_1 = defaultdict(list)
block_2 = defaultdict(list)

for idx, name in enumerate(source_names):

    block_1[first4_len(name)].append(idx)
    block_2[first3_last3(name)].append(idx)

print("Block indexes ready.")


# ------------------------------------------------------------
# 4. Evaluate the 300 uncovered names
# ------------------------------------------------------------

results = []

for _, row in uncovered_sample.iterrows():

    query = str(row["business_name_clean"])

    # -----------------------------
    # Block 1
    # -----------------------------
    candidates_1 = block_1.get(
        first4_len(query),
        []
    )

    # -----------------------------
    # Block 2
    # -----------------------------
    candidates_2 = block_2.get(
        first3_last3(query),
        []
    )

    # -----------------------------
    # Score candidates
    # -----------------------------
    best_1 = 0
    best_2 = 0

    best_name_1 = ""
    best_name_2 = ""

    for idx in candidates_1:

        candidate = source_names[idx]

        score = fuzz.ratio(query, candidate)

        if score > best_1:
            best_1 = score
            best_name_1 = candidate

    for idx in candidates_2:

        candidate = source_names[idx]

        score = fuzz.ratio(query, candidate)

        if score > best_2:
            best_2 = score
            best_name_2 = candidate

    results.append({
        "source1_entity_id": row["entity_id"],
        "s1_name": query,

        "block1_candidates": len(candidates_1),
        "block1_best_score": best_1,
        "block1_best_name": best_name_1,

        "block2_candidates": len(candidates_2),
        "block2_best_score": best_2,
        "block2_best_name": best_name_2
    })


results_df = pd.DataFrame(results)


# ------------------------------------------------------------
# 5. Summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("BLOCKING SUMMARY")
print("=" * 70)

for block in ["block1", "block2"]:

    print(f"\n{block.upper()}")

    print(
        "Average candidates:",
        round(
            results_df[f"{block}_candidates"].mean(),
            2
        )
    )

    print(
        "Median candidates:",
        results_df[f"{block}_candidates"].median()
    )

    print(
        "95th percentile:",
        results_df[f"{block}_candidates"].quantile(.95)
    )

    print(
        "Maximum:",
        results_df[f"{block}_candidates"].max()
    )

    print(
        "Best score >= 90:",
        (
            results_df[f"{block}_best_score"] >= 90
        ).sum(),
        "/ 300"
    )

    print(
        "Best score >= 95:",
        (
            results_df[f"{block}_best_score"] >= 95
        ).sum(),
        "/ 300"
    )


# ------------------------------------------------------------
# 6. Show examples
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SAMPLE RESULTS")
print("=" * 70)

print(
    results_df[
        [
            "s1_name",
            "block1_candidates",
            "block1_best_score",
            "block1_best_name",
            "block2_candidates",
            "block2_best_score",
            "block2_best_name"
        ]
    ]
    .head(30)
    .to_string(index=False)
)

print("\n" + "=" * 70)
print("EXPERIMENT COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# STEP 3J — CANONICAL BUSINESS-NAME BLOCKING TEST
# ============================================================

import pandas as pd
import re
import time

print("=" * 70)
print("CANONICAL BUSINESS-NAME BLOCKING TEST")
print("=" * 70)


# ------------------------------------------------------------
# Canonicalization function
# ------------------------------------------------------------

def canonical_name(name):

    if not name:
        return ""

    x = str(name).lower().strip()

    # Common legal/business suffixes
    suffix_patterns = [
        r"privatelimited$",
        r"privateltd$",
        r"private$",
        r"pvtlimited$",
        r"pvtltd$",
        r"pvt$",
        r"limited$",
        r"ltd$",
        r"llp$",
        r"llc$",
        r"incorporated$",
        r"inc$",
        r"corporation$",
        r"corp$",
        r"company$",
        r"co$",
        r"plc$",
        r"sarl$",
        r"eurl$",
        r"gmbh$",
        r"sa$",
        r"sas$",
    ]

    # Remove repeated suffixes
    changed = True

    while changed:
        old = x

        for pattern in suffix_patterns:
            x = re.sub(pattern, "", x)

        changed = (x != old)

    # Common internal business abbreviations
    replacements = {
        "privatelimited": "private",
        "privateltd": "private",
        "pvtlimited": "private",
        "pvtltd": "private",
    }

    for old, new in replacements.items():
        x = x.replace(old, new)

    return x.strip()


# ------------------------------------------------------------
# Test the 300 S1 names
# ------------------------------------------------------------

test_names = uncovered_sample[
    ["entity_id", "business_name_clean"]
].copy()

test_names["canonical_name"] = (
    test_names["business_name_clean"]
    .apply(canonical_name)
)

print("\nSample canonicalization:")

print(
    test_names[
        ["business_name_clean", "canonical_name"]
    ]
    .head(30)
    .to_string(index=False)
)


# ------------------------------------------------------------
# Build canonical source-name index
# ------------------------------------------------------------

print("\nBuilding canonical source-name index...")

start = time.time()

canonical_source = names_df.copy()

canonical_source["canonical_name"] = (
    canonical_source["business_name_clean"]
    .apply(canonical_name)
)

canonical_source = canonical_source[
    canonical_source["canonical_name"] != ""
]

canonical_groups = (
    canonical_source
    .groupby("canonical_name")
    .size()
)

print(
    f"Canonical source rows: "
    f"{len(canonical_source):,}"
)

print(
    f"Unique canonical names: "
    f"{canonical_groups.shape[0]:,}"
)

print(
    f"Index build time: "
    f"{(time.time()-start)/60:.2f} minutes"
)


# ------------------------------------------------------------
# Candidate counts for the 300 sample
# ------------------------------------------------------------

source_by_canonical = (
    canonical_source
    .groupby("canonical_name")
    .size()
    .to_dict()
)

test_names["canonical_candidates"] = (
    test_names["canonical_name"]
    .map(source_by_canonical)
    .fillna(0)
    .astype(int)
)


# ------------------------------------------------------------
# Show distribution
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CANONICAL BLOCK SUMMARY")
print("=" * 70)

print(
    "Average candidates:",
    round(
        test_names["canonical_candidates"].mean(),
        2
    )
)

print(
    "Median candidates:",
    test_names["canonical_candidates"].median()
)

print(
    "95th percentile:",
    test_names["canonical_candidates"].quantile(.95)
)

print(
    "Maximum:",
    test_names["canonical_candidates"].max()
)

print(
    "S1 with >=1 canonical candidate:",
    (
        test_names["canonical_candidates"] > 0
    ).sum(),
    "/ 300"
)


# ------------------------------------------------------------
# Show examples
# ------------------------------------------------------------

print("\nEXAMPLES:")

print(
    test_names[
        [
            "business_name_clean",
            "canonical_name",
            "canonical_candidates"
        ]
    ]
    .head(40)
    .to_string(index=False)
)

print("\n" + "=" * 70)
print("CANONICAL BLOCK TEST COMPLETE")
print("=" * 70)
```

```python
# Canonical-name normalization
LEGAL_SUFFIXES = [
    "privatelimited", "privateltd", "private",
    "pvtlimited", "pvtltd", "pvt",
    "limited", "ltd", "llp", "llc",
    "incorporated", "inc",
    "corporation", "corp",
    "company", "co",
    "plc", "sarl", "eurl",
    "gmbh", "sas", "sa"
]

def canonical_name(s):
    s = str(s).lower().strip()

    # Keep only letters/numbers
    s = re.sub(r"[^a-z0-9]", "", s)

    # Remove repeated legal suffixes
    changed = True
    while changed and s:
        changed = False
        for suffix in LEGAL_SUFFIXES:
            if s.endswith(suffix) and len(s) > len(suffix) + 2:
                s = s[:-len(suffix)]
                changed = True
                break

    return s


print("Creating canonical names...")

train_s2["canonical_name"] = train_s2["business_name_clean"].map(canonical_name)
train_s3["canonical_name"] = train_s3["business_name_clean"].map(canonical_name)

# Remove empty canonical names
s2_valid = train_s2[train_s2["canonical_name"] != ""]
s3_valid = train_s3[train_s3["canonical_name"] != ""]

print("Valid S2 names:", len(s2_valid))
print("Valid S3 names:", len(s3_valid))

# Build canonical -> entity IDs
canonical_index = defaultdict(list)

for eid, cname in zip(
    s2_valid["entity_id"].values,
    s2_valid["canonical_name"].values
):
    canonical_index[cname].append(eid)

for eid, cname in zip(
    s3_valid["entity_id"].values,
    s3_valid["canonical_name"].values
):
    canonical_index[cname].append(eid)

print("Unique canonical keys:", len(canonical_index))

# Show candidate-size distribution
candidate_sizes = [len(v) for v in canonical_index.values()]

print("\nCanonical candidate distribution:")
print("Mean :", round(sum(candidate_sizes) / len(candidate_sizes), 2))
print("Median:", sorted(candidate_sizes)[len(candidate_sizes)//2])
print("Max   :", max(candidate_sizes))

print("\nIndex built successfully.")
```

```python
import time

print("Evaluating exact-name + exact-address blocking...")
t0 = time.time()

# Create a compact ground-truth table inside DuckDB
db.execute("DROP TABLE IF EXISTS gt_eval")

db.execute("""
CREATE TABLE gt_eval AS
SELECT
    source1_entity_id,
    matched_entity_ids
FROM read_csv_auto(
    '/content/drive/MyDrive/dataset/train_ground_truth.tsv',
    delim='\\t',
    header=true,
    all_varchar=true
)
""")

# Explode the comma-separated true matches
db.execute("DROP TABLE IF EXISTS true_pairs")

db.execute("""
CREATE TABLE true_pairs AS
SELECT
    source1_entity_id,
    trim(match_id) AS matched_entity_id
FROM gt_eval,
UNNEST(string_split(matched_entity_ids, ',')) AS t(match_id)
WHERE trim(match_id) <> ''
""")

print("Ground-truth pairs:",
      db.execute("SELECT COUNT(*) FROM true_pairs").fetchone()[0])

# Evaluate whether every true pair can be produced by:
#   1. exact business_name_clean
#   2. exact business_address_clean
#
# We only test true pairs, so we don't materialize the millions
# of non-matching candidate pairs.

query = """
WITH s1 AS (
    SELECT
        entity_id,
        business_name_clean,
        business_address_clean
    FROM read_csv_auto(
        '/content/drive/MyDrive/dataset/cleaned_train_source1.tsv',
        delim='\\t',
        header=true,
        all_varchar=true
    )
),

matched AS (
    SELECT
        tp.source1_entity_id,
        tp.matched_entity_id,
        s1.business_name_clean AS s1_name,
        s1.business_address_clean AS s1_address,
        src.business_name_clean AS src_name,
        src.business_address_clean AS src_address
    FROM true_pairs tp
    JOIN s1
        ON s1.entity_id = tp.source1_entity_id
    JOIN sources src
        ON src.entity_id = tp.matched_entity_id
)

SELECT
    COUNT(*) AS total_true_pairs,

    SUM(
        CASE
            WHEN
                (s1_name <> '' AND s1_name = src_name)
                OR
                (s1_address <> '' AND s1_address = src_address)
            THEN 1
            ELSE 0
        END
    ) AS retrieved_true_pairs,

    COUNT(DISTINCT source1_entity_id) AS s1_with_true_pairs,

    COUNT(DISTINCT
        CASE
            WHEN
                (s1_name <> '' AND s1_name = src_name)
                OR
                (s1_address <> '' AND s1_address = src_address)
            THEN source1_entity_id
        END
    ) AS s1_with_at_least_one_hit

FROM matched
"""

result = db.execute(query).fetchone()

total_pairs = result[0]
retrieved_pairs = result[1]
total_s1 = result[2]
hit_s1 = result[3]

print("\n" + "=" * 65)
print("A+B EXACT BLOCKING — TRAINING GROUND-TRUTH RECALL")
print("=" * 65)

print(f"Total true pairs        : {total_pairs:,}")
print(f"Retrieved true pairs    : {retrieved_pairs:,}")
print(f"Pair recall             : {retrieved_pairs / total_pairs:.6%}")

print(f"\nS1 entities with matches : {total_s1:,}")
print(f"S1 entities with ≥1 hit  : {hit_s1:,}")
print(f"Entity recall            : {hit_s1 / total_s1:.6%}")

print(f"\nTime: {time.time()-t0:.1f}s")
print("=" * 65)
```

```python
import time

print("Profiling missing true pairs...")
t0 = time.time()

query = """
WITH s1 AS (
    SELECT
        entity_id,
        lower(regexp_replace(coalesce(business_name_clean, ''), '[^a-z0-9]', '', 'g')) AS name,
        lower(regexp_replace(coalesce(business_address_clean, ''), '[^a-z0-9]', '', 'g')) AS address
    FROM read_csv_auto(
        '/content/drive/MyDrive/dataset/cleaned_train_source1.tsv',
        delim='\\t',
        header=true,
        all_varchar=true
    )
),

matched AS (
    SELECT
        tp.source1_entity_id,
        tp.matched_entity_id,

        s1.name AS s1_name,
        src.business_name_clean AS src_name_raw,

        s1.address AS s1_address,
        src.business_address_clean AS src_address_raw
    FROM true_pairs tp
    JOIN s1
        ON s1.entity_id = tp.source1_entity_id
    JOIN sources src
        ON src.entity_id = tp.matched_entity_id
),

x AS (
    SELECT
        *,

        lower(
            regexp_replace(
                coalesce(src_name_raw, ''),
                '[^a-z0-9]',
                '',
                'g'
            )
        ) AS src_name,

        lower(
            regexp_replace(
                coalesce(src_address_raw, ''),
                '[^a-z0-9]',
                '',
                'g'
            )
        ) AS src_address

    FROM matched
),

features AS (
    SELECT
        *,

        -- Existing exact routes
        s1_name = src_name AS exact_name,
        s1_address = src_address AS exact_address,

        -- Name prefixes
        (
            length(s1_name) >= 4
            AND length(src_name) >= 4
            AND left(s1_name, 4) = left(src_name, 4)
        ) AS name_prefix4,

        (
            length(s1_name) >= 3
            AND length(src_name) >= 3
            AND left(s1_name, 3) = left(src_name, 3)
        ) AS name_prefix3,

        -- Prefix + suffix
        (
            length(s1_name) >= 6
            AND length(src_name) >= 6
            AND left(s1_name, 3) = left(src_name, 3)
            AND right(s1_name, 3) = right(src_name, 3)
        ) AS name_prefix3_suffix3,

        -- Similar name length
        (
            abs(length(s1_name) - length(src_name)) <= 3
        ) AS name_len_diff3,

        (
            abs(length(s1_name) - length(src_name)) <= 5
        ) AS name_len_diff5,

        -- Extract address numbers and test overlap
        regexp_extract_all(s1_address, '[0-9]+') AS s1_numbers,
        regexp_extract_all(src_address, '[0-9]+') AS src_numbers

    FROM x
),

final AS (
    SELECT
        *,
        list_has_any(s1_numbers, src_numbers) AS address_number_overlap
    FROM features
)

SELECT
    COUNT(*) AS total_pairs,

    SUM(CASE WHEN exact_name OR exact_address THEN 1 ELSE 0 END)
        AS exact_ab,

    SUM(CASE WHEN name_prefix4 THEN 1 ELSE 0 END)
        AS prefix4,

    SUM(CASE WHEN name_prefix3_suffix3 THEN 1 ELSE 0 END)
        AS prefix3_suffix3,

    SUM(CASE WHEN name_prefix4 AND name_len_diff3 THEN 1 ELSE 0 END)
        AS prefix4_len3,

    SUM(CASE WHEN name_prefix3_suffix3 AND name_len_diff5 THEN 1 ELSE 0 END)
        AS prefix3suffix3_len5,

    SUM(CASE WHEN address_number_overlap THEN 1 ELSE 0 END)
        AS address_number,

    SUM(
        CASE
            WHEN name_prefix4
             AND name_len_diff5
             AND address_number_overlap
            THEN 1 ELSE 0
        END
    ) AS prefix4_len5_number,

    SUM(
        CASE
            WHEN name_prefix3_suffix3
             AND name_len_diff5
             AND address_number_overlap
            THEN 1 ELSE 0
        END
    ) AS prefix3suffix3_len5_number

FROM final
"""

r = db.execute(query).fetchone()

labels = [
    "Total true pairs",
    "Exact name OR address",
    "Name prefix-4",
    "Name prefix-3 + suffix-3",
    "Prefix-4 + name length ±3",
    "Prefix3+suffix3 + name length ±5",
    "Address number overlap",
    "Prefix4 + length ±5 + number",
    "Prefix3+suffix3 + length ±5 + number"
]

total = r[0]

print("\n" + "=" * 70)
print("MISSING-PAIR BLOCKING PROFILE")
print("=" * 70)

for label, value in zip(labels, r):
    if label == "Total true pairs":
        print(f"{label:45s}: {value:,}")
    else:
        print(f"{label:45s}: {value:,}  ({value/total:.2%})")

print(f"\nTime: {time.time()-t0:.1f}s")
print("=" * 70)
```

```python
import time

print("Measuring union blocking recall...")
t0 = time.time()

query = """
WITH s1 AS (
    SELECT
        entity_id,
        lower(
            regexp_replace(
                coalesce(business_name_clean, ''),
                '[^a-z0-9]',
                '',
                'g'
            )
        ) AS name,
        lower(
            regexp_replace(
                coalesce(business_address_clean, ''),
                '[^a-z0-9]',
                '',
                'g'
            )
        ) AS address
    FROM read_csv_auto(
        '/content/drive/MyDrive/dataset/cleaned_train_source1.tsv',
        delim='\\t',
        header=true,
        all_varchar=true
    )
),

matched AS (
    SELECT
        tp.source1_entity_id,
        tp.matched_entity_id,
        s1.name AS s1_name,
        s1.address AS s1_address,
        lower(
            regexp_replace(
                coalesce(src.business_name_clean, ''),
                '[^a-z0-9]',
                '',
                'g'
            )
        ) AS src_name,
        lower(
            regexp_replace(
                coalesce(src.business_address_clean, ''),
                '[^a-z0-9]',
                '',
                'g'
            )
        ) AS src_address
    FROM true_pairs tp
    JOIN s1
        ON s1.entity_id = tp.source1_entity_id
    JOIN sources src
        ON src.entity_id = tp.matched_entity_id
),

signals AS (
    SELECT
        *,

        (
            s1_name <> ''
            AND s1_name = src_name
        ) AS exact_name,

        (
            s1_address <> ''
            AND s1_address = src_address
        ) AS exact_address,

        (
            length(s1_name) >= 4
            AND length(src_name) >= 4
            AND left(s1_name, 4) = left(src_name, 4)
        ) AS prefix4,

        list_has_any(
            regexp_extract_all(s1_address, '[0-9]+'),
            regexp_extract_all(src_address, '[0-9]+')
        ) AS number_overlap

    FROM matched
)

SELECT
    COUNT(*) AS total_pairs,

    SUM(
        CASE
            WHEN exact_name
              OR exact_address
              OR prefix4
              OR number_overlap
            THEN 1 ELSE 0
        END
    ) AS union_pairs,

    SUM(
        CASE
            WHEN prefix4 OR number_overlap
            THEN 1 ELSE 0
        END
    ) AS prefix_or_number,

    SUM(
        CASE
            WHEN prefix4 AND number_overlap
            THEN 1 ELSE 0
        END
    ) AS prefix_and_number

FROM signals
"""

r = db.execute(query).fetchone()

total = r[0]
union_pairs = r[1]
prefix_or_number = r[2]
prefix_and_number = r[3]

print("\n" + "=" * 70)
print("BLOCKING UNION RECALL")
print("=" * 70)

print(f"Total true pairs                 : {total:,}")
print(
    f"Prefix-4 OR number OR exact AB  : "
    f"{union_pairs:,} ({union_pairs/total:.2%})"
)
print(
    f"Prefix-4 OR number              : "
    f"{prefix_or_number:,} ({prefix_or_number/total:.2%})"
)
print(
    f"Prefix-4 AND number              : "
    f"{prefix_and_number:,} ({prefix_and_number/total:.2%})"
)

print(f"\nTime: {time.time()-t0:.1f}s")
print("=" * 70)
```

```python
import time

print("Testing tighter blocking combinations on known true pairs...")
t0 = time.time()

query = """
WITH s1 AS (
    SELECT
        entity_id,
        lower(regexp_replace(coalesce(business_name_clean, ''), '[^a-z0-9]', '', 'g')) AS name,
        lower(regexp_replace(coalesce(business_address_clean, ''), '[^a-z0-9]', '', 'g')) AS address
    FROM read_csv_auto(
        '/content/drive/MyDrive/dataset/cleaned_train_source1.tsv',
        delim='\\t',
        header=true,
        all_varchar=true
    )
),

matched AS (
    SELECT
        tp.source1_entity_id,
        tp.matched_entity_id,
        s1.name AS s1_name,
        s1.address AS s1_address,
        lower(regexp_replace(coalesce(src.business_name_clean, ''), '[^a-z0-9]', '', 'g')) AS src_name,
        lower(regexp_replace(coalesce(src.business_address_clean, ''), '[^a-z0-9]', '', 'g')) AS src_address
    FROM true_pairs tp
    JOIN s1 ON s1.entity_id = tp.source1_entity_id
    JOIN sources src ON src.entity_id = tp.matched_entity_id
),

f AS (
    SELECT
        *,
        (
            s1_name <> ''
            AND s1_name = src_name
        ) AS exact_name,

        (
            s1_address <> ''
            AND s1_address = src_address
        ) AS exact_address,

        (
            length(s1_name) >= 4
            AND length(src_name) >= 4
            AND left(s1_name, 4) = left(src_name, 4)
        ) AS prefix4,

        (
            abs(length(s1_name) - length(src_name)) <= 3
        ) AS len3,

        (
            abs(length(s1_name) - length(src_name)) <= 5
        ) AS len5,

        list_has_any(
            regexp_extract_all(s1_address, '[0-9]+'),
            regexp_extract_all(src_address, '[0-9]+')
        ) AS num_overlap

    FROM matched
)

SELECT
    COUNT(*) AS total,

    SUM(CASE
        WHEN exact_name OR exact_address
             OR (prefix4 AND len3)
             OR num_overlap
        THEN 1 ELSE 0 END) AS r1,

    SUM(CASE
        WHEN exact_name OR exact_address
             OR (prefix4 AND len5)
             OR num_overlap
        THEN 1 ELSE 0 END) AS r2,

    SUM(CASE
        WHEN exact_name OR exact_address
             OR (prefix4 AND len3 AND num_overlap)
             OR num_overlap
        THEN 1 ELSE 0 END) AS r3,

    SUM(CASE
        WHEN exact_name OR exact_address
             OR (prefix4 AND len5 AND num_overlap)
             OR num_overlap
        THEN 1 ELSE 0 END) AS r4,

    SUM(CASE
        WHEN exact_name OR exact_address
             OR (prefix4 AND len3)
             OR (prefix4 AND len5)
             OR num_overlap
        THEN 1 ELSE 0 END) AS r5

FROM f
"""

r = db.execute(query).fetchone()
total = r[0]

print("\n" + "=" * 70)
print("TIGHT BLOCKING RECALL")
print("=" * 70)

labels = [
    "Prefix4 + length±3 OR number",
    "Prefix4 + length±5 OR number",
    "Prefix4 + length±3 + number OR number",
    "Prefix4 + length±5 + number OR number",
    "All tested routes combined"
]

for label, value in zip(labels, r[1:]):
    print(f"{label:45s}: {value:,}  ({value/total:.2%})")

print(f"\nTime: {time.time()-t0:.1f}s")
print("=" * 70)
```

```python
import time

print("Building compact blocking-key tables...")
t0 = time.time()

# Reopen writable connection
db.close()

db = duckdb.connect(TRAIN_DB)
db.execute("SET memory_limit='4GB'")
db.execute("SET threads=2")
db.execute("SET preserve_insertion_order=false")

# Source blocking keys
db.execute("DROP TABLE IF EXISTS source_block_keys")

db.execute("""
CREATE TABLE source_block_keys AS
SELECT
    entity_id,
    lower(
        regexp_replace(
            coalesce(business_name_clean, ''),
            '[^a-z0-9]',
            '',
            'g'
        )
    ) AS name,
    CASE
        WHEN length(
            regexp_replace(
                coalesce(business_name_clean, ''),
                '[^a-z0-9]',
                '',
                'g'
            )
        ) >= 4
        THEN left(
            regexp_replace(
                coalesce(business_name_clean, ''),
                '[^a-z0-9]',
                '',
                'g'
            ),
            4
        )
        ELSE ''
    END AS prefix4,

    CASE
        WHEN length(
            regexp_replace(
                coalesce(business_name_clean, ''),
                '[^a-z0-9]',
                '',
                'g'
            )
        ) >= 1
        THEN CAST(
            floor(
                length(
                    regexp_replace(
                        coalesce(business_name_clean, ''),
                        '[^a-z0-9]',
                        '',
                        'g'
                    )
                ) / 5
            ) AS INTEGER
        )
        ELSE -1
    END AS len_bucket

FROM sources
WHERE business_name_clean IS NOT NULL
""")

print(f"Source keys built: {time.time()-t0:.1f}s")

db.execute("""
CREATE INDEX idx_source_prefix_len
ON source_block_keys(prefix4, len_bucket)
""")

print(f"Source block index built: {time.time()-t0:.1f}s")

# Basic key statistics
stats = db.execute("""
SELECT
    COUNT(*) AS rows,
    COUNT(DISTINCT prefix4 || ':' || CAST(len_bucket AS VARCHAR))
        AS unique_blocks
FROM source_block_keys
WHERE prefix4 <> ''
""").fetchone()

print("\nSource rows:", f"{stats[0]:,}")
print("Unique prefix+length blocks:", f"{stats[1]:,}")

print(f"\nTotal time: {time.time()-t0:.1f}s")
!free -h
```

```python
import time

print("Testing selective recovery signals...")
t0 = time.time()

query = """
WITH s1 AS (
    SELECT
        entity_id,
        lower(regexp_replace(coalesce(business_name_clean, ''), '[^a-z0-9]', '', 'g')) AS name,
        lower(regexp_replace(coalesce(business_address_clean, ''), '[^a-z0-9]', '', 'g')) AS address
    FROM read_csv_auto(
        '/content/drive/MyDrive/dataset/cleaned_train_source1.tsv',
        delim='\\t',
        header=true,
        all_varchar=true
    )
),

matched AS (
    SELECT
        tp.source1_entity_id,
        s1.name AS s1_name,
        s1.address AS s1_address,

        lower(regexp_replace(
            coalesce(src.business_name_clean, ''),
            '[^a-z0-9]', '', 'g'
        )) AS src_name,

        lower(regexp_replace(
            coalesce(src.business_address_clean, ''),
            '[^a-z0-9]', '', 'g'
        )) AS src_address

    FROM true_pairs tp
    JOIN s1
      ON s1.entity_id = tp.source1_entity_id
    JOIN sources src
      ON src.entity_id = tp.matched_entity_id
),

f AS (
    SELECT
        *,
        length(s1_name) AS l1,
        length(src_name) AS l2,
        left(s1_name, 4) AS p4_1,
        left(src_name, 4) AS p4_2,
        right(s1_name, 3) AS s3_1,
        right(src_name, 3) AS s3_2,

        list_has_any(
            regexp_extract_all(s1_address, '[0-9]+'),
            regexp_extract_all(src_address, '[0-9]+')
        ) AS num

    FROM matched
)

SELECT
    COUNT(*) AS total,

    SUM(CASE
        WHEN p4_1 = p4_2
         AND abs(l1-l2) <= 2
        THEN 1 ELSE 0 END) AS p4_len2,

    SUM(CASE
        WHEN p4_1 = p4_2
         AND s3_1 = s3_2
         AND abs(l1-l2) <= 5
        THEN 1 ELSE 0 END) AS p4_suffix3_len5,

    SUM(CASE
        WHEN p4_1 = p4_2
         AND num
         AND abs(l1-l2) <= 5
        THEN 1 ELSE 0 END) AS p4_num_len5,

    SUM(CASE
        WHEN p4_1 = p4_2
         AND num
        THEN 1 ELSE 0 END) AS p4_num,

    SUM(CASE
        WHEN p4_1 = p4_2
         AND s3_1 = s3_2
         AND num
        THEN 1 ELSE 0 END) AS p4_suffix_num

FROM f
"""

r = db.execute(query).fetchone()
total = r[0]

print("\n" + "=" * 70)
print("SELECTIVE RECOVERY RECALL")
print("=" * 70)

labels = [
    "Prefix4 + length ±2",
    "Prefix4 + suffix3 + length ±5",
    "Prefix4 + number + length ±5",
    "Prefix4 + number",
    "Prefix4 + suffix3 + number"
]

for label, value in zip(labels, r[1:]):
    print(f"{label:45s}: {value:,} ({value/total:.2%})")

print(f"\nTime: {time.time()-t0:.1f}s")
print("=" * 70)
```

```python
import duckdb

DB = "/content/train_recall_v2.db"

cx = duckdb.connect(DB, read_only=True)
cx.execute("SET memory_limit='4GB'")
cx.execute("SET threads=2")

# Strip common legal/business suffixes repeatedly.
# This mirrors the canonical-name logic we tested previously.
suffix_pattern = (
    "(privatelimited|privateltd|private|pvtlimited|pvtltd|pvt|"
    "limited|ltd|llp|llc|incorporated|inc|corporation|corp|"
    "company|co|plc|sarl|eurl|gmbh|sas|sa)$"
)

canonical_expr = f"""
trim(
    regexp_replace(
        regexp_replace(
            regexp_replace(
                regexp_replace(
                    regexp_replace(
                        lower(coalesce(business_name_clean, '')),
                        '{suffix_pattern}', ''
                    ),
                    '{suffix_pattern}', ''
                ),
                '{suffix_pattern}', ''
            ),
            '{suffix_pattern}', ''
        ),
        '{suffix_pattern}', ''
    )
)
"""

# Create temporary canonical source table.
print("Building temporary canonical source table...")

cx.execute("DROP TABLE IF EXISTS canonical_sources")

cx.execute(f"""
CREATE TEMP TABLE canonical_sources AS
SELECT
    entity_id,
    {canonical_expr} AS canonical_name
FROM sources
WHERE business_name_clean IS NOT NULL
  AND trim(business_name_clean) <> ''
""")

print(
    cx.execute("""
        SELECT
            COUNT(*) AS rows,
            COUNT(DISTINCT canonical_name) AS unique_canonical_names
        FROM canonical_sources
        WHERE canonical_name <> ''
    """).fetchdf().to_string(index=False)
)

# Canonical key frequency.
print("\nCanonical key frequency:")
print(
    cx.execute("""
        SELECT
            AVG(cnt) AS avg_sources_per_key,
            MEDIAN(cnt) AS median_sources_per_key,
            MAX(cnt) AS max_sources_per_key
        FROM (
            SELECT canonical_name, COUNT(*) AS cnt
            FROM canonical_sources
            WHERE canonical_name <> ''
            GROUP BY canonical_name
        )
    """).fetchdf().to_string(index=False)
)

# Measure canonical recall directly against true pairs.
print("\nEvaluating canonical recall against ALL true pairs...")

result = cx.execute(f"""
WITH true_with_canonical AS (
    SELECT
        tp.source1_entity_id,
        tp.matched_entity_id,
        cs1.canonical_name AS s1_canonical,
        cs2.canonical_name AS match_canonical
    FROM true_pairs tp
    JOIN s1_block_keys s1
      ON s1.entity_id = tp.source1_entity_id
    JOIN canonical_sources cs1
      ON cs1.entity_id = tp.source1_entity_id
    JOIN canonical_sources cs2
      ON cs2.entity_id = tp.matched_entity_id
),
canonical_hits AS (
    SELECT *
    FROM true_with_canonical
    WHERE s1_canonical <> ''
      AND s1_canonical = match_canonical
)
SELECT
    (SELECT COUNT(*) FROM true_pairs) AS total_true_pairs,
    COUNT(*) AS canonical_true_pairs,
    ROUND(
        100.0 * COUNT(*) /
        (SELECT COUNT(*) FROM true_pairs),
        4
    ) AS canonical_pair_recall_pct,
    COUNT(DISTINCT source1_entity_id) AS s1_with_canonical_hit,
    ROUND(
        100.0 * COUNT(DISTINCT source1_entity_id) /
        (SELECT COUNT(DISTINCT source1_entity_id) FROM true_pairs),
        4
    ) AS canonical_entity_recall_pct
FROM canonical_hits
""").fetchdf()

print(result.to_string(index=False))
```

```python
import lightgbm as lgb
import numpy as np
import pandas as pd

MODEL_PATH = "/content/drive/MyDrive/amazon_training/models/entity_match_lightgbm_corrected.txt"

feature_cols = [
    "name_exact",
    "name_clean_exact",
    "name_ratio",
    "name_token_sort",
    "name_token_set",
    "address_exact",
    "address_clean_exact",
    "address_ratio",
    "address_token_sort",
    "address_token_set",
    "number_overlap",
    "name_length_diff",
    "address_length_diff",
    "country_match",
    "source_type",
    "name_translit_ratio",
    "name_translit_token_sort",
    "name_translit_token_set",
    "name_script_diff"
]

model = lgb.Booster(model_file=MODEL_PATH)

X_valid = valid_df[feature_cols]
y_valid = valid_df["label"].values

valid_df = valid_df.copy()
valid_df["score"] = model.predict(X_valid)

print("Validation score summary")
print("=" * 60)
print(valid_df["score"].describe(percentiles=[.01,.05,.10,.25,.50,.75,.90,.95,.99]).to_string())

print("\nPositive vs negative score distribution")
print("=" * 60)

print(
    valid_df.groupby("label")["score"]
    .describe(percentiles=[.01,.05,.10,.25,.50,.75,.90,.95,.99])
    .to_string()
)

print("\nThreshold precision/recall")
print("=" * 60)

rows = []

for threshold in [0.50, 0.70, 0.80, 0.90, 0.95, 0.97, 0.98, 0.99, 0.995, 0.999]:
    pred = valid_df["score"].values >= threshold

    tp = np.sum((pred == 1) & (y_valid == 1))
    fp = np.sum((pred == 1) & (y_valid == 0))
    fn = np.sum((pred == 0) & (y_valid == 1))

    precision = tp / (tp + fp) if tp + fp else 0
    recall = tp / (tp + fn) if tp + fn else 0

    beta2 = 0.5 ** 2
    f05 = (
        (1 + beta2) * precision * recall /
        (beta2 * precision + recall)
        if precision + recall else 0
    )

    rows.append([
        threshold,
        int(pred.sum()),
        precision,
        recall,
        f05
    ])

print(
    pd.DataFrame(
        rows,
        columns=["threshold", "predicted_pairs", "precision", "recall", "F0.5"]
    ).to_string(index=False)
)
```

```python
import duckdb

S2_SCORE = "/content/drive/MyDrive/amazon_training/features/test_scores_corrected_source2_v2.parquet"
S3_SCORE = "/content/drive/MyDrive/amazon_training/features/test_scores_corrected_source3_v2.parquet"

cx = duckdb.connect()
cx.execute("SET memory_limit='4GB'")
cx.execute("SET threads=2")

query = f"""
WITH all_scores AS (
    SELECT * FROM read_parquet('{S2_SCORE}')
    UNION ALL
    SELECT * FROM read_parquet('{S3_SCORE}')
),
ranked AS (
    SELECT
        *,
        ROW_NUMBER() OVER (
            PARTITION BY source1_entity_id
            ORDER BY score DESC, matched_entity_id
        ) AS rn,
        LEAD(score) OVER (
            PARTITION BY source1_entity_id
            ORDER BY score DESC, matched_entity_id
        ) AS next_score
    FROM all_scores
),
agg AS (
    SELECT
        source1_entity_id,
        MAX(CASE WHEN rn = 1 THEN score END) AS top1,
        MAX(CASE WHEN rn = 2 THEN score END) AS top2,
        MAX(CASE WHEN rn = 1 THEN matched_entity_id END) AS top1_id,
        MAX(CASE WHEN rn = 2 THEN matched_entity_id END) AS top2_id,
        MAX(CASE WHEN rn = 1 THEN matched_source END) AS top1_source,
        MAX(CASE WHEN rn = 2 THEN matched_source END) AS top2_source,
        COUNT(*) AS candidate_count
    FROM ranked
    GROUP BY source1_entity_id
)

SELECT
    COUNT(*) AS covered_s1,

    SUM(
        CASE WHEN top1 >= 0.80
        THEN 1 ELSE 0 END
    ) AS top1_score80_s1,

    SUM(
        CASE WHEN top1 >= 0.90
        THEN 1 ELSE 0 END
    ) AS top1_score90_s1,

    SUM(
        CASE WHEN top1 >= 0.80
          AND top2 IS NOT NULL
          AND top2 >= 0.80
        THEN 2 ELSE 1 END
    ) AS pairs_top2_score80,

    SUM(
        CASE
            WHEN top1 >= 0.80
             AND top2 IS NOT NULL
             AND top1 - top2 <= 0.0001
            THEN 2
            WHEN top1 >= 0.80
            THEN 1
            ELSE 0
        END
    ) AS pairs_gap_0001,

    SUM(
        CASE
            WHEN top1 >= 0.80
             AND top2 IS NOT NULL
             AND top1 - top2 <= 0.001
            THEN 2
            WHEN top1 >= 0.80
            THEN 1
            ELSE 0
        END
    ) AS pairs_gap_001,

    SUM(
        CASE
            WHEN top1 >= 0.80
             AND top2 IS NOT NULL
             AND top1 - top2 <= 0.01
            THEN 2
            WHEN top1 >= 0.80
            THEN 1
            ELSE 0
        END
    ) AS pairs_gap_01

FROM agg;
"""

result = cx.execute(query).fetchdf()

print(result.to_string(index=False))
```

```python
import numpy as np
import pandas as pd

df = valid_df.copy()

# Rank candidates within each S1 by model score
df["rank"] = (
    df.groupby("source1_entity_id")["score"]
      .rank(method="first", ascending=False)
      .astype(int)
)

# Each S1 can have multiple candidates in the validation sample.
# Evaluate several selection rules.
def evaluate_rule(df, rule_name, selector):
    selected = selector(df)

    # Aggregate TP/FP/FN per S1
    stats = []

    for s1, group in df.groupby("source1_entity_id"):
        pred_ids = set(
            selected.loc[
                selected["source1_entity_id"] == s1,
                "matched_entity_id"
            ]
        )

        true_ids = set(
            group.loc[group["label"] == 1, "matched_entity_id"]
        )

        tp = len(pred_ids & true_ids)
        fp = len(pred_ids - true_ids)
        fn = len(true_ids - pred_ids)

        precision = tp / (tp + fp) if tp + fp else 0.0
        recall = tp / (tp + fn) if tp + fn else 0.0

        beta2 = 0.5 ** 2
        f05 = (
            (1 + beta2) * precision * recall /
            (beta2 * precision + recall)
            if precision + recall > 0
            else 0.0
        )

        stats.append((precision, recall, f05))

    stats = np.array(stats)

    return {
        "rule": rule_name,
        "S1": len(stats),
        "predicted_pairs": len(selected),
        "macro_precision": stats[:, 0].mean(),
        "macro_recall": stats[:, 1].mean(),
        "macro_F0.5": stats[:, 2].mean()
    }


rules = []

# Rule 1: top 1
rules.append(
    evaluate_rule(
        df,
        "Top-1",
        lambda x: x[x["rank"] == 1]
    )
)

# Rule 2: top 2 if score gap <= threshold
for gap in [0.0001, 0.001, 0.01]:
    def make_selector(g):
        def selector(x):
            first = x[x["rank"] == 1].copy()

            second = x[x["rank"] == 2].copy()

            # Match second candidate back to its S1's top score
            top_scores = (
                first[["source1_entity_id", "score"]]
                .rename(columns={"score": "top_score"})
            )

            second = second.merge(
                top_scores,
                on="source1_entity_id",
                how="left"
            )

            second = second[
                (second["top_score"] - second["score"]) <= g
            ]

            return pd.concat([first, second], ignore_index=True)

        return selector

    rules.append(
        evaluate_rule(
            df,
            f"Top-2 if gap <= {gap}",
            make_selector(gap)
        )
    )

result = pd.DataFrame(rules)

print(result.to_string(index=False))
```

```python
import duckdb

TEMP_MATCHING = "/content/drive/MyDrive/amazon_training_v2/submission/matching_results_gap001_temp.tsv"

ROUTE_A = "/content/drive/MyDrive/amazon_training_v2/candidates/route_a_name.parquet"
ROUTE_B = "/content/drive/MyDrive/amazon_training_v2/candidates/route_b_address.parquet"

cx3 = duckdb.connect()
cx3.execute("SET memory_limit='4GB'")
cx3.execute("SET threads=2")

# ---------------------------------------------------------
# Load candidate union
# ---------------------------------------------------------
print("Building candidate union...")

cx3.execute(f"""
CREATE OR REPLACE TEMP TABLE candidates AS

SELECT DISTINCT
    source1_entity_id,
    matched_entity_id,
    'source2' AS matched_source
FROM read_parquet('{ROUTE_A}')

UNION

SELECT DISTINCT
    source1_entity_id,
    matched_entity_id,
    'source3' AS matched_source
FROM read_parquet('{ROUTE_B}')
""")

print(
    cx3.execute("""
        SELECT
            COUNT(*) AS candidate_pairs,
            COUNT(DISTINCT source1_entity_id) AS candidate_s1
        FROM candidates
    """).fetchdf().to_string(index=False)
)

# ---------------------------------------------------------
# Parse predictions into individual pairs
# ---------------------------------------------------------
cx3.execute(f"""
CREATE OR REPLACE TEMP TABLE predictions AS
SELECT
    source1_entity_id,
    trim(matched_entity_id) AS matched_entity_id
FROM read_csv(
    '{TEMP_MATCHING}',
    delim='\\t',
    header=true,
    columns={{
        'source1_entity_id':'VARCHAR',
        'matched_entity_ids':'VARCHAR'
    }}
),
UNNEST(
    string_split(
        COALESCE(matched_entity_ids, ''),
        ','
    )
) AS t(matched_entity_id)
WHERE trim(matched_entity_id) <> ''
""")

# ---------------------------------------------------------
# Prediction statistics
# ---------------------------------------------------------
print("\nPREDICTION STATISTICS")
print("=" * 60)

print(
    cx3.execute("""
        SELECT
            COUNT(*) AS predicted_pairs,
            COUNT(DISTINCT source1_entity_id) AS predicted_s1,
            COUNT(DISTINCT source1_entity_id || '|' || matched_entity_id)
                AS unique_pairs
        FROM predictions
    """).fetchdf().to_string(index=False)
)

# ---------------------------------------------------------
# Check duplicate predicted pairs
# ---------------------------------------------------------
duplicates = cx3.execute("""
SELECT COUNT(*) AS duplicate_extra_rows
FROM (
    SELECT
        source1_entity_id,
        matched_entity_id,
        COUNT(*) AS cnt
    FROM predictions
    GROUP BY source1_entity_id, matched_entity_id
    HAVING COUNT(*) > 1
)
""").fetchone()[0]

print("\nDUPLICATE PAIR GROUPS:", duplicates)

# ---------------------------------------------------------
# Check every prediction exists in candidate set
# ---------------------------------------------------------
missing = cx3.execute("""
SELECT COUNT(*) AS missing_predictions
FROM predictions p
LEFT JOIN candidates c
    ON p.source1_entity_id = c.source1_entity_id
   AND p.matched_entity_id = c.matched_entity_id
WHERE c.matched_entity_id IS NULL
""").fetchone()[0]

print("PREDICTIONS MISSING FROM CANDIDATES:", missing)

# ---------------------------------------------------------
# Explicitly count blank/covered S1s
# ---------------------------------------------------------
s1_check = cx3.execute(f"""
SELECT
    COUNT(*) AS total_s1,
    SUM(
        CASE
            WHEN matched_entity_ids IS NULL
              OR trim(matched_entity_ids) = ''
            THEN 1 ELSE 0
        END
    ) AS blank_s1,
    SUM(
        CASE
            WHEN matched_entity_ids IS NOT NULL
             AND trim(matched_entity_ids) <> ''
            THEN 1 ELSE 0
        END
    ) AS nonblank_s1
FROM read_csv(
    '{TEMP_MATCHING}',
    delim='\\t',
    header=true,
    columns={{
        'source1_entity_id':'VARCHAR',
        'matched_entity_ids':'VARCHAR'
    }}
)
""").fetchdf()

print("\nS1 COVERAGE")
print("=" * 60)
print(s1_check.to_string(index=False))

# ---------------------------------------------------------
# Final status
# ---------------------------------------------------------
print("\n" + "=" * 60)

if missing == 0 and duplicates == 0:
    print("PASS: Every prediction belongs to the candidate set.")
    print("PASS: No duplicate prediction pairs.")
else:
    print("FAIL: DO NOT SUBMIT.")
```
